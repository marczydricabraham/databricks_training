```sql
WITH CustomerRanks AS (
    SELECT 
        total_spend,
        -- This assigns a decimal from 0 to 1 based on spend rank
        PERCENT_RANK() OVER (ORDER BY total_spend DESC) as percentile
    FROM capstone_101.gold.high_value_customers
),
Totals AS (
    SELECT 
        SUM(total_spend) as grand_total,
        SUM(CASE WHEN percentile <= 0.01 THEN total_spend ELSE 0 END) as top_tier_total
    FROM CustomerRanks
)
SELECT 
    'Top 1% Customers' as Category,
    ROUND(top_tier_total, 2) as Tier_Total,
    ROUND((top_tier_total / grand_total) * 100, 2) as Percent_of_Revenue
FROM Totals;
```
