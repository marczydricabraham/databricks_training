```sql
-- Use this for a Map or Bar Chart
SELECT 
    Country, 
    ROUND(total_revenue, 2) as Revenue,
    -- Calculate share of total to show dominance
    ROUND((total_revenue / SUM(total_revenue) OVER()) * 100, 2) as Market_Share_Percent
FROM capstone_101.gold.total_revenue_by_country
ORDER BY total_revenue DESC;
```
